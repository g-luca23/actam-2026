### C3 · A page assembled from pieces

**Core · Combine functions**

Write `paragraph(text)`, returning a `<p>...</p>` element.

Use `heading` and `paragraph` to build a page fragment with one heading and two paragraphs. Render the combined string once.

Then write `short_page(title, first, second)` so another call can produce another page.

````python
def paragraph(text):
    return "<p>" + text + "</p>"

def short_page(title, first, second):
    return heading(title)+paragraph(first)+paragraph(second)

display(HTML(short_page("title","paragrafo1", "paragrafo2")))
````

### Text that contains HTML characters

Suppose the heading itself should say `<hello>`. Putting that text directly inside our tags can make the renderer treat it as markup.

Python supplies `escape` to represent special characters as text. Look at both the string and the rendered output:

````python
from html import escape

print(escape("<hello> & goodbye"))
display(HTML("<h2>" + escape("<hello> & goodbye") + "</h2>"))
````

### C4 · A heading that preserves its text

**Core · Adapt a function**

Write `safe_heading(text)` and `safe_paragraph(text)` using `escape` on the text inside the element.

Render `"Ada & Bruno"` and `"<hello>"`. Both should appear exactly as text.

Why should we escape the content, rather than the entire string including the tags? Use these safe functions in later projects.

````python
def safe_heading(text):
    return escape(heading(text))
    
    
def safe_paragraph(text):
    return escape(paragraph(text))

display(HTML(safe_heading("title>>>")+safe_paragraph("paragraph><><")))

````

### D1 · One list item

**Core · Build one piece**

HTML uses `<li>...</li>` for a list item.

Write `item(text)`, returning one list item with escaped text. Try `"cat"` and `"salt & pepper"`.

Now use a comprehension on `["cat", "fox", "owl"]` to build a **list of three HTML strings**. Inspect that list before rendering anything.

````python
def item(animal):
    return "<li>" + escape(animal) + "</li>"

items = [item(animal) for animal in ["cat", "bear", "pear"]]
print(items)
print("".join(items))
display(HTML("".join(items)))
````
